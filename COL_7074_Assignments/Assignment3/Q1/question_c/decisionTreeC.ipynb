```python
import pandas as pd
import numpy as np
```

```python
train_data = pd.read_csv("data/train.csv")
validation_data = pd.read_csv("data/validation.csv")
test_data = pd.read_csv("data/test.csv")

print(f"Shape of train_data is : {train_data.shape}")
print(f"Shape of validation_data is : {validation_data.shape}")
print(f"Shape of test_data is : {test_data.shape}")
train_data.head()
```

```python
from utils import convertCategoricalColumnsToOneHotEncoding, pre_process

train_data_encoded, val_data_encoded, test_data_encoded = convertCategoricalColumnsToOneHotEncoding(train_data, validation_data, test_data)
# Step 2: Encode the columns["Unnamed", "team", "opp", "host", "month"]
continuous_cols = ["Unnamed", "year", "fow", "score", "rpo"]

trainDataX, trainDataY, validationDataX, validationDataY, testDataX, testDataY =  pre_process(train_data_encoded, val_data_encoded, test_data_encoded, categorical_cols = [], continuous_cols = continuous_cols)
trainDataX.head()
```

```python
# Question C

# now build a decision tree to predict the result given the trainingData
# Step 0: 
from decisionTree import DecisionTreeModel
from utils import evaluate, plotA
from prunningUtils import post_prune_greedy
import matplotlib.pyplot as plt

maxDepths = [15, 25, 35, 45]
trainAccuracies = []
validationAccuracies = []
testAccuracies = []

new_categorical_cols = list(trainDataX.columns[trainDataX.dtypes == bool])
new_categorical_cols.extend(["toss", "bat_first", "format"])

print(f"Currrent length of categorical columns : {len(new_categorical_cols)}")

categorical_cols = new_categorical_cols
continuous_cols = ["Unnamed", "year", "fow", "score", "rpo"]

for maxDepth in maxDepths:
    print(f"---------------Testing for maxDepth : {maxDepth}-------------------")
    model = DecisionTreeModel(maxDepth, categorical_cols = categorical_cols, continuous_cols = continuous_cols)

    # TRAIN THE MODEL WITH TRAIN DATA
    print("Training on the train data -- Building the full tree")
    model.fit(trainDataX, trainDataY)

    print("Once the full tree is being build lets start by prunning the internal nodes that gives us the maxm accuracy")
    history = post_prune_greedy(model, trainDataX, trainDataY, validationDataX, validationDataY, testDataX, testDataY)
    nodes, train_accs, val_accs, test_accs = zip(*history)

    plotA(nodes, train_accs, val_accs, test_accs, xLabel = "Number of Nodes in Tree", title = f"Post-Pruning Curve (maxDepth={maxDepth})")
```

Observations:

Training & validation accuracy steadily increase with depth — as expected, deeper trees fit more complex patterns.

Test accuracy remains nearly constant (~0.594) → this indicates overfitting beyond depth ~15–25, since the model keeps improving on train/val but not generalizing better.

The number of pruned nodes stabilizes (~45–50), showing the pruning process effectively trims redundant complexity even for deeper initial trees.

Depth 25 or 35 seems a balanced trade-off — good validation accuracy (≈0.72–0.73) without unnecessary model size.

At depth 45, the tree:

Became more complex initially, but then pruning cut off more aggressively (47 nodes pruned vs. 52 before).
As a result, it’s effectively a simpler model post-pruning — hence the drop in training accuracy.
The small bump in test accuracy (0.5936 → 0.5946) is likely just random variation — not a real improvement — because both train and validation accuracies fell.
This typically means pruning may have over-pruned a few useful branches, sacrificing some fit quality.

Conclusion:
Going deeper than ~25–35 doesn’t improve test performance; it only adds complexity. You might cap your tree depth around 25–30 for optimal generalization and efficiency.
