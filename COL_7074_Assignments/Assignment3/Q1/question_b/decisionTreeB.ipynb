```python
import pandas as pd
import numpy as np
```

```python
train_data = pd.read_csv("../data/train.csv")
validation_data = pd.read_csv("../data/validation.csv")
test_data = pd.read_csv("../data/test.csv")

print(f"Shape of train_data is : {train_data.shape}")
print(f"Shape of validation_data is : {validation_data.shape}")
print(f"Shape of test_data is : {test_data.shape}")
train_data.head()
```

```python
import sys
import os
sys.path.insert(0, os.path.abspath('..'))  # Adjust path as needed
from utils import convertCategoricalColumnsToOneHotEncoding

train_data_encoded, val_data_encoded, test_data_encoded = convertCategoricalColumnsToOneHotEncoding(train_data, validation_data, test_data)
```

```python
from utils import pre_process

# Step 2: Encode the columns["Unnamed", "team", "opp", "host", "month"]
continuous_cols = ["Unnamed", "year", "fow", "score", "rpo"]

trainDataX, trainDataY, validationDataX, validationDataY, testDataX, testDataY =  pre_process(train_data_encoded, val_data_encoded, test_data_encoded, categorical_cols = [], continuous_cols = continuous_cols)
trainDataX.head()
```

```python
new_categorical_cols = list(trainDataX.columns[trainDataX.dtypes == bool])
new_categorical_cols.extend(["toss", "bat_first", "format"])

print(f"Currrent length of categorical columns : {len(new_categorical_cols)}")
```

```python
# Question B

# now build a decision tree to predict the result given the trainingData
# Step 0: 
from decisionTree import DecisionTreeModel
from utils import evaluate, plotA

maxDepths = [15, 25, 35, 45]
trainAccuracies = []
validationAccuracies = []
testAccuracies = []

categorical_cols = new_categorical_cols
continuous_cols = ["Unnamed", "year", "fow", "score", "rpo"]

for maxDepth in maxDepths:
    print(f"---------------Testing for maxDepth : {maxDepth}-------------------")
    model = DecisionTreeModel(maxDepth, categorical_cols = categorical_cols, continuous_cols = continuous_cols)

    # TRAIN THE MODEL WITH TRAIN DATA
    print("Training on the train data")
    model.fit(trainDataX, trainDataY)

    # if maxDepth == 5:
    #     model.print_tree()
    
    # PREDICT THE MODEL ON TRAIN DATA
    print("Prediction on train data")
    predictY_train = model.predict(trainDataX)
    
    # EVALUATION
    train_accuracy = evaluate(predictY_train, trainDataY)
    
    # PREDICT THE MODEL ON VALIDATION DATA
    print("Prediction on validation data")
    predictY_validation = model.predict(validationDataX)
    
    # EVALUATION
    validation_accuracy = evaluate(predictY_validation, validationDataY)
    
    # PREDICT THE MODEL ON Test DATA
    print("Prediction on test data")
    predictY_test = model.predict(testDataX)
    
    # EVALUATION
    test_accuracy = evaluate(predictY_test, testDataY)

    trainAccuracies.append(train_accuracy)
    validationAccuracies.append(validation_accuracy)
    testAccuracies.append(test_accuracy)

plotA(maxDepths, trainAccuracies, validationAccuracies, testAccuracies)
```

| Max Depth | Train Accuracy | Validation Accuracy | Test Accuracy |
| --------- | -------------- | ------------------- | ------------- |
| 15        | 0.788          | 0.614               | 0.608         |
| 25        | 0.936          | 0.616               | 0.620         |
| 35        | 0.982          | 0.584               | 0.626         |
| 45        | 0.991          | 0.579               | 0.625         |

Training accuracy increases steadily as the maximum depth grows (from 0.78 → 0.99).
→ This shows the model is fitting the training data better as depth increases — a clear sign of overfitting.

Validation and test accuracies remain almost constant (around 0.58–0.63).
→ Even though the tree becomes more complex, it doesn’t generalize better on unseen data.

One-Hot Encoding slightly improves test performance compared to part (a).

In part (a): Test accuracy peaked at ~0.592

In part (b): Test accuracy improved to ~0.626
 Encoding categorical variables helped the model represent discrete features more effectively.

Beyond depth = 35, there’s no gain in validation/test accuracy → optimal depth seems to be around 30–35.

After applying one-hot encoding, the decision tree achieved slightly better generalization, with test accuracy increasing from ~0.59 (without encoding) to ~0.62. This improvement arises because one-hot encoding allows the model to better distinguish between categorical levels. However, increasing tree depth beyond 35 leads to overfitting — the training accuracy reaches nearly 100%, while test accuracy stagnates — indicating that deeper trees memorize the training data rather than learning generalizable patterns.
