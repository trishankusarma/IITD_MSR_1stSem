```python
# evaluation_notebook.py

import pandas as pd
import os
from utils import evaluate

# Path to test labels
test_labels_path = "data/test.csv"

# Folder containing prediction files
pred_folder = "outputs"

# Load actual test labels
test_data = pd.read_csv(test_labels_path)
y_true = test_data["result"]

# List all prediction CSV files and sort in ascending order
pred_files = sorted([f for f in os.listdir(pred_folder) if f.endswith(".csv")])

# Evaluate each prediction file
results = []

for file in pred_files:
    pred_path = os.path.join(pred_folder, file)
    preds = pd.read_csv(pred_path)["result"]
    
    print(f"Evaluating {file} ...")
    acc = evaluate(preds, y_true)
    
    results.append({
        "file": file,
        "accuracy": acc
    })

# Optional: create a summary DataFrame
summary_df = pd.DataFrame(results)
print("\nSummary of all predictions in ascending order:")
display(summary_df)
```
