```python
from utils.utils import load_images_from_folders, plotF1Scores  
from neural_network import NeuralNetwork
import logging

logging.info("Loading training data...")
train_X, train_y = load_images_from_folders("dataset/train")
logging.info(f"Training samples: {train_X.shape[0]}")

logging.info("Loading test data...")
test_X, test_y = load_images_from_folders("dataset/test")
logging.info(f"Test samples: {test_X.shape[0]}")
```

```python
M = 32
n = 32 * 32 * 3
r = 36
learning_rate = 0.01
epochs = 400
batch_size = M
```

```python
layer = [512, 256, 128, 64]

logging.info(f"\nTraining Part d model with {layer} layer ...")
nn = NeuralNetwork(hidden_layers=layer, learning_rate=learning_rate, input_size=n, output_size=r)
nn.fit(train_X, train_y, epochs=epochs, batch_size=batch_size, use_relu=True)
```

```python
import numpy as np
np.savez("consonant_model_weights.npz", **nn.params)
```
